```python
# ============================================================
# EXACT CLASS-WISE IMAGE COUNT
# Searches ONLY Urban_Dataset and all its subfolders
# ============================================================

import os
from collections import Counter

URBAN_DATASET = "/content/drive/MyDrive/Urban_Dataset"

image_extensions = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".bmp",
    ".avif"
)

classes = [
    "Residential",
    "Commercial",
    "Construction",
    "School"
]

class_counts = Counter()
unclassified = []

for root, dirs, files in os.walk(URBAN_DATASET):

    for file in files:

        if not file.lower().endswith(image_extensions):
            continue

        # Full path
        full_path = os.path.join(root, file)

        # Check which class folder the image belongs to
        matched_class = None

        for cls in classes:
            if cls.lower() in full_path.lower():
                matched_class = cls
                break

        if matched_class:
            class_counts[matched_class] += 1
        else:
            unclassified.append(full_path)


# ============================================================
# PRINT RESULTS
# ============================================================

print("\n" + "=" * 60)
print("EXACT CLASS-WISE IMAGE COUNT")
print("=" * 60)

for cls in classes:
    print(f"{cls:<20}: {class_counts[cls]} images")

print("-" * 60)

total_classified = sum(class_counts.values())

print(f"{'CLASSIFIED TOTAL':<20}: {total_classified} images")
print(f"{'UNCLASSIFIED':<20}: {len(unclassified)} images")
print(f"{'TOTAL FOUND':<20}: {total_classified + len(unclassified)} images")

print("=" * 60)
```

```python
# ============================================================
# STEP 1: STRATIFIED TRAIN / VALIDATION / TEST SPLIT
# Urban Dataset
# ============================================================

import os
import random
import shutil
from collections import defaultdict

# ------------------------------------------------------------
# SETTINGS
# ------------------------------------------------------------

SOURCE_DIR = "/content/drive/MyDrive/Urban_Dataset"

OUTPUT_DIR = "/content/drive/MyDrive/Urban_Dataset_Split"

CLASSES = [
    "Residential",
    "Commercial",
    "Construction",
    "School"
]

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".bmp",
    ".avif"
)

SEED = 42

TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15

random.seed(SEED)


# ------------------------------------------------------------
# COLLECT IMAGES CLASS-WISE
# ------------------------------------------------------------

class_images = {}

for cls in CLASSES:

    class_dir = os.path.join(SOURCE_DIR, cls)

    images = []

    for root, dirs, files in os.walk(class_dir):

        for file in files:

            if file.lower().endswith(IMAGE_EXTENSIONS):

                images.append(
                    os.path.join(root, file)
                )

    class_images[cls] = images


# ------------------------------------------------------------
# DISPLAY ORIGINAL DATASET
# ------------------------------------------------------------

print("=" * 70)
print("ORIGINAL DATASET")
print("=" * 70)

total_images = 0

for cls in CLASSES:

    count = len(class_images[cls])

    total_images += count

    print(f"{cls:<15}: {count}")

print("-" * 70)
print(f"{'TOTAL':<15}: {total_images}")
print("=" * 70)


# ------------------------------------------------------------
# CREATE OUTPUT DIRECTORIES
# ------------------------------------------------------------

for split in ["train", "validation", "test"]:

    for cls in CLASSES:

        os.makedirs(
            os.path.join(
                OUTPUT_DIR,
                split,
                cls
            ),
            exist_ok=True
        )


# ------------------------------------------------------------
# STRATIFIED SPLIT
# ------------------------------------------------------------

split_counts = defaultdict(dict)

for cls in CLASSES:

    images = class_images[cls].copy()

    random.shuffle(images)

    n = len(images)

    train_end = int(n * TRAIN_RATIO)

    val_end = train_end + int(n * VAL_RATIO)

    train_images = images[:train_end]

    val_images = images[train_end:val_end]

    test_images = images[val_end:]

    split_data = {
        "train": train_images,
        "validation": val_images,
        "test": test_images
    }

    # --------------------------------------------------------
    # COPY IMAGES
    # --------------------------------------------------------

    for split, files_list in split_data.items():

        split_counts[split][cls] = len(files_list)

        destination_dir = os.path.join(
            OUTPUT_DIR,
            split,
            cls
        )

        for source_file in files_list:

            destination_file = os.path.join(
                destination_dir,
                os.path.basename(source_file)
            )

            shutil.copy2(
                source_file,
                destination_file
            )


# ------------------------------------------------------------
# DISPLAY FINAL SPLIT
# ------------------------------------------------------------

print("\n")
print("=" * 70)
print("FINAL STRATIFIED DATASET SPLIT")
print("=" * 70)

for split in ["train", "validation", "test"]:

    print(f"\n{split.upper()}")
    print("-" * 70)

    split_total = 0

    for cls in CLASSES:

        count = split_counts[split][cls]

        split_total += count

        print(f"{cls:<15}: {count}")

    print("-" * 70)
    print(f"{'TOTAL':<15}: {split_total}")


# ------------------------------------------------------------
# FINAL TOTAL
# ------------------------------------------------------------

grand_total = sum(
    sum(split_counts[split].values())
    for split in ["train", "validation", "test"]
)

print("\n")
print("=" * 70)
print("FINAL DATASET SUMMARY")
print("=" * 70)

print(f"Original images : {total_images}")
print(f"Split images    : {grand_total}")
print(f"Difference      : {total_images - grand_total}")

print("\nDataset saved at:")
print(OUTPUT_DIR)

print("\nRandom seed:", SEED)
print("=" * 70)
```

```python
# ============================================================
# STEP 2: DATASET QUALITY AND IMAGE CHARACTERISTICS ANALYSIS
# ============================================================

import os
import hashlib
from PIL import Image
from collections import Counter, defaultdict

# ------------------------------------------------------------
# DATASET PATH
# ------------------------------------------------------------

DATASET_DIR = "/content/drive/MyDrive/Urban_Dataset"

CLASSES = [
    "Residential",
    "Commercial",
    "Construction",
    "School"
]

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".bmp",
    ".avif"
)


# ------------------------------------------------------------
# STORAGE
# ------------------------------------------------------------

class_stats = defaultdict(lambda: {
    "count": 0,
    "widths": [],
    "heights": [],
    "ratios": [],
    "modes": Counter(),
    "formats": Counter()
})

all_hashes = defaultdict(list)

corrupted_images = []

total_images = 0


# ------------------------------------------------------------
# SCAN DATASET
# ------------------------------------------------------------

for cls in CLASSES:

    class_dir = os.path.join(DATASET_DIR, cls)

    if not os.path.exists(class_dir):

        print(f"WARNING: {class_dir} does not exist")
        continue

    for root, dirs, files in os.walk(class_dir):

        for filename in files:

            if not filename.lower().endswith(IMAGE_EXTENSIONS):
                continue

            filepath = os.path.join(root, filename)

            total_images += 1

            try:

                # ------------------------------------------------
                # OPEN IMAGE
                # ------------------------------------------------

                with Image.open(filepath) as img:

                    # Verify image integrity
                    img.verify()

                # Reopen after verify
                with Image.open(filepath) as img:

                    width, height = img.size

                    mode = img.mode

                    image_format = img.format

                # ------------------------------------------------
                # STATISTICS
                # ------------------------------------------------

                class_stats[cls]["count"] += 1

                class_stats[cls]["widths"].append(width)

                class_stats[cls]["heights"].append(height)

                class_stats[cls]["ratios"].append(
                    width / height
                )

                class_stats[cls]["modes"][mode] += 1

                class_stats[cls]["formats"][image_format] += 1


                # ------------------------------------------------
                # IMAGE HASH FOR DUPLICATE DETECTION
                # ------------------------------------------------

                with open(filepath, "rb") as f:

                    file_hash = hashlib.md5(
                        f.read()
                    ).hexdigest()

                all_hashes[file_hash].append(filepath)


            except Exception as e:

                corrupted_images.append(
                    (filepath, str(e))
                )


# ============================================================
# DATASET SUMMARY
# ============================================================

print("\n")
print("=" * 75)
print("DATASET QUALITY ANALYSIS")
print("=" * 75)

print(f"\nTotal images scanned : {total_images}")
print(f"Corrupted images     : {len(corrupted_images)}")


# ============================================================
# CLASS-WISE ANALYSIS
# ============================================================

print("\n")
print("=" * 75)
print("CLASS-WISE IMAGE STATISTICS")
print("=" * 75)

for cls in CLASSES:

    stats = class_stats[cls]

    widths = stats["widths"]
    heights = stats["heights"]
    ratios = stats["ratios"]

    if len(widths) == 0:
        continue

    print(f"\n{cls}")
    print("-" * 75)

    print(f"Images              : {stats['count']}")

    print(
        f"Width               : "
        f"min={min(widths)}, "
        f"max={max(widths)}, "
        f"mean={sum(widths)/len(widths):.2f}"
    )

    print(
        f"Height              : "
        f"min={min(heights)}, "
        f"max={max(heights)}, "
        f"mean={sum(heights)/len(heights):.2f}"
    )

    print(
        f"Aspect ratio        : "
        f"min={min(ratios):.3f}, "
        f"max={max(ratios):.3f}, "
        f"mean={sum(ratios)/len(ratios):.3f}"
    )

    print(
        f"Color modes         : "
        f"{dict(stats['modes'])}"
    )

    print(
        f"File formats        : "
        f"{dict(stats['formats'])}"
    )


# ============================================================
# DUPLICATE ANALYSIS
# ============================================================

duplicate_groups = {
    h: paths
    for h, paths in all_hashes.items()
    if len(paths) > 1
}

duplicate_images = sum(
    len(paths) - 1
    for paths in duplicate_groups.values()
)


print("\n")
print("=" * 75)
print("DUPLICATE IMAGE ANALYSIS")
print("=" * 75)

print(
    f"Unique image hashes      : {len(all_hashes)}"
)

print(
    f"Duplicate groups         : {len(duplicate_groups)}"
)

print(
    f"Duplicate copies        : {duplicate_images}"
)


# ============================================================
# SHOW DUPLICATE FILES
# ============================================================

if duplicate_groups:

    print("\nDuplicate groups found:")

    group_number = 1

    for h, paths in duplicate_groups.items():

        print(f"\nGroup {group_number}")

        for path in paths:

            print(path)

        group_number += 1

        if group_number > 10:

            print("\nShowing first 10 duplicate groups only.")

            break

else:

    print("\nNo exact duplicate images found.")


# ============================================================
# CORRUPTED IMAGE REPORT
# ============================================================

print("\n")
print("=" * 75)
print("CORRUPTED IMAGE ANALYSIS")
print("=" * 75)

if corrupted_images:

    print(
        f"Corrupted/unreadable images: "
        f"{len(corrupted_images)}"
    )

    print("\nExamples:")

    for filepath, error in corrupted_images[:10]:

        print(filepath)
        print("Error:", error)

else:

    print("No corrupted or unreadable images detected.")


# ============================================================
# FINAL DATASET VALIDATION
# ============================================================

valid_images = total_images - len(corrupted_images)

print("\n")
print("=" * 75)
print("FINAL DATASET VALIDATION")
print("=" * 75)

print(f"Total scanned       : {total_images}")
print(f"Valid images        : {valid_images}")
print(f"Corrupted images    : {len(corrupted_images)}")
print(f"Exact duplicates    : {duplicate_images}")

print("=" * 75)
```

```python
# ============================================================
# STEP 3: CHECK DUPLICATES ACROSS TRAIN / VALIDATION / TEST
# ============================================================

import os
import hashlib
from collections import defaultdict

SPLIT_DIR = "/content/drive/MyDrive/Urban_Dataset_Split"

SPLITS = [
    "train",
    "validation",
    "test"
]

CLASSES = [
    "Residential",
    "Commercial",
    "Construction",
    "School"
]

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".bmp",
    ".avif"
)

# ------------------------------------------------------------
# FIND HASHES IN EACH SPLIT
# ------------------------------------------------------------

hash_locations = defaultdict(list)

for split in SPLITS:

    split_dir = os.path.join(SPLIT_DIR, split)

    for root, dirs, files in os.walk(split_dir):

        for filename in files:

            if not filename.lower().endswith(IMAGE_EXTENSIONS):
                continue

            filepath = os.path.join(root, filename)

            try:

                with open(filepath, "rb") as f:

                    file_hash = hashlib.md5(
                        f.read()
                    ).hexdigest()

                hash_locations[file_hash].append(
                    (split, filepath)
                )

            except Exception as e:

                print("Could not read:", filepath)
                print("Error:", e)


# ------------------------------------------------------------
# FIND DUPLICATES
# ------------------------------------------------------------

duplicate_groups = {
    h: locations
    for h, locations in hash_locations.items()
    if len(locations) > 1
}


# ------------------------------------------------------------
# CLASSIFY DUPLICATES
# ------------------------------------------------------------

cross_split_groups = {}

within_split_groups = {}

for h, locations in duplicate_groups.items():

    splits_present = set(
        location[0]
        for location in locations
    )

    if len(splits_present) > 1:

        cross_split_groups[h] = locations

    else:

        within_split_groups[h] = locations


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("DUPLICATE LEAKAGE ANALYSIS")
print("=" * 75)

print(
    f"Total duplicate groups       : "
    f"{len(duplicate_groups)}"
)

print(
    f"Cross-split duplicate groups : "
    f"{len(cross_split_groups)}"
)

print(
    f"Within-split duplicate groups: "
    f"{len(within_split_groups)}"
)


# ------------------------------------------------------------
# SHOW CROSS-SPLIT DUPLICATES
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("CROSS-SPLIT DUPLICATES")
print("=" * 75)

if cross_split_groups:

    group_no = 1

    for h, locations in cross_split_groups.items():

        print(f"\nGroup {group_no}")

        for split, filepath in locations:

            print(f"[{split.upper()}]")
            print(filepath)

        group_no += 1

else:

    print(
        "GOOD: No exact duplicates were found "
        "across train, validation and test."
    )


# ------------------------------------------------------------
# SHOW WITHIN-SPLIT DUPLICATES
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("WITHIN-SPLIT DUPLICATES")
print("=" * 75)

if within_split_groups:

    group_no = 1

    for h, locations in within_split_groups.items():

        print(f"\nGroup {group_no}")

        for split, filepath in locations:

            print(f"[{split.upper()}]")
            print(filepath)

        group_no += 1

else:

    print("No within-split duplicate groups found.")


# ------------------------------------------------------------
# FINAL DECISION
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("DATA LEAKAGE DECISION")
print("=" * 75)

if cross_split_groups:

    print(
        "WARNING: Cross-split duplicate leakage detected."
    )

    print(
        "The dataset split should be regenerated after "
        "removing/grouping duplicate images."
    )

else:

    print(
        "PASS: No exact duplicate leakage detected "
        "between train, validation and test."
    )

print("=" * 75)
```

```python
# ============================================================
# STEP 4: REMOVE EXACT DUPLICATES + CREATE CLEAN SPLIT
# ============================================================

import os
import shutil
import hashlib
import random
from collections import defaultdict

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

SOURCE_DIR = "/content/drive/MyDrive/Urban_Dataset"

CLEAN_DIR = "/content/drive/MyDrive/Urban_Dataset_Clean"

SPLIT_DIR = "/content/drive/MyDrive/Urban_Dataset_Clean_Split"

CLASSES = [
    "Residential",
    "Commercial",
    "Construction",
    "School"
]

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".webp",
    ".bmp",
    ".avif"
)

SEED = 42

TRAIN_RATIO = 0.70
VAL_RATIO = 0.15
TEST_RATIO = 0.15


# ------------------------------------------------------------
# REMOVE PREVIOUS CLEAN OUTPUTS IF THEY EXIST
# ------------------------------------------------------------

for directory in [CLEAN_DIR, SPLIT_DIR]:

    if os.path.exists(directory):

        shutil.rmtree(directory)

        print("Removed previous:", directory)


# ------------------------------------------------------------
# CREATE CLEAN DATASET DIRECTORIES
# ------------------------------------------------------------

for cls in CLASSES:

    os.makedirs(
        os.path.join(CLEAN_DIR, cls),
        exist_ok=True
    )


# ------------------------------------------------------------
# FIND ALL IMAGES AND GROUP BY HASH
# ------------------------------------------------------------

hash_groups = defaultdict(list)

original_counts = defaultdict(int)

print("=" * 75)
print("SCANNING ORIGINAL DATASET")
print("=" * 75)

for cls in CLASSES:

    class_dir = os.path.join(
        SOURCE_DIR,
        cls
    )

    for root, dirs, files in os.walk(class_dir):

        for filename in files:

            if not filename.lower().endswith(
                IMAGE_EXTENSIONS
            ):
                continue

            filepath = os.path.join(
                root,
                filename
            )

            original_counts[cls] += 1

            # ------------------------------------------------
            # MD5 HASH
            # ------------------------------------------------

            with open(filepath, "rb") as f:

                file_hash = hashlib.md5(
                    f.read()
                ).hexdigest()

            hash_groups[file_hash].append(
                (cls, filepath)
            )


# ------------------------------------------------------------
# KEEP ONE COPY OF EACH UNIQUE IMAGE
# ------------------------------------------------------------

unique_files = defaultdict(list)

duplicate_files = []

for file_hash, files_list in hash_groups.items():

    # Keep first copy
    cls, first_file = files_list[0]

    unique_files[cls].append(
        first_file
    )

    # Remaining copies are duplicates
    if len(files_list) > 1:

        for duplicate in files_list[1:]:

            duplicate_files.append(
                duplicate
            )


# ------------------------------------------------------------
# COPY UNIQUE IMAGES TO CLEAN DATASET
# ------------------------------------------------------------

for cls in CLASSES:

    for source_file in unique_files[cls]:

        filename = os.path.basename(
            source_file
        )

        destination = os.path.join(
            CLEAN_DIR,
            cls,
            filename
        )

        shutil.copy2(
            source_file,
            destination
        )


# ------------------------------------------------------------
# DISPLAY DUPLICATE REMOVAL RESULTS
# ------------------------------------------------------------

print("\n")
print("=" * 75)
print("DUPLICATE REMOVAL RESULTS")
print("=" * 75)

original_total = 0
unique_total = 0

for cls in CLASSES:

    original_count = original_counts[cls]

    unique_count = len(
        unique_files[cls]
    )

    removed = (
        original_count -
        unique_count
    )

    original_total += original_count
    unique_total += unique_count

    print(
        f"{cls:<15}: "
        f"Original={original_count:<4} "
        f"Unique={unique_count:<4} "
        f"Removed={removed}"
    )

print("-" * 75)

print(
    f"{'TOTAL':<15}: "
    f"Original={original_total:<4} "
    f"Unique={unique_total:<4} "
    f"Removed={len(duplicate_files)}"
)


# ------------------------------------------------------------
# CREATE TRAIN / VALIDATION / TEST DIRECTORIES
# ------------------------------------------------------------

for split in [
    "train",
    "validation",
    "test"
]:

    for cls in CLASSES:

        os.makedirs(
            os.path.join(
                SPLIT_DIR,
                split,
                cls
            ),
            exist_ok=True
        )


# ------------------------------------------------------------
# STRATIFIED SPLIT
# ------------------------------------------------------------

random.seed(SEED)

split_counts = defaultdict(dict)

for cls in CLASSES:

    images = unique_files[cls].copy()

    random.shuffle(images)

    n = len(images)

    train_end = int(
        n * TRAIN_RATIO
    )

    val_end = (
        train_end +
        int(n * VAL_RATIO)
    )

    train_images = images[:train_end]

    validation_images = images[
        train_end:val_end
    ]

    test_images = images[
        val_end:
    ]

    split_data = {
        "train": train_images,
        "validation": validation_images,
        "test": test_images
    }

    # --------------------------------------------------------
    # COPY SPLIT FILES
    # --------------------------------------------------------

    for split, files_list in split_data.items():

        split_counts[split][cls] = len(
            files_list
        )

        destination_dir = os.path.join(
            SPLIT_DIR,
            split,
            cls
        )

        for source_file in files_list:

            destination_file = os.path.join(
                destination_dir,
                os.path.basename(source_file)
            )

            shutil.copy2(
                source_file,
                destination_file
            )


# ============================================================
# PRINT FINAL SPLIT
# ============================================================

print("\n")
print("=" * 75)
print("CLEAN DATASET SPLIT")
print("=" * 75)

grand_total = 0

for split in [
    "train",
    "validation",
    "test"
]:

    print(f"\n{split.upper()}")
    print("-" * 75)

    split_total = 0

    for cls in CLASSES:

        count = split_counts[split][cls]

        split_total += count

        print(
            f"{cls:<15}: {count}"
        )

    print("-" * 75)

    print(
        f"{'TOTAL':<15}: {split_total}"
    )

    grand_total += split_total


# ============================================================
# FINAL VALIDATION
# ============================================================

print("\n")
print("=" * 75)
print("FINAL DATASET VALIDATION")
print("=" * 75)

print(
    "Original images       :",
    original_total
)

print(
    "Duplicate copies      :",
    len(duplicate_files)
)

print(
    "Unique images         :",
    unique_total
)

print(
    "Images in final split :",
    grand_total
)

print(
    "Difference            :",
    unique_total - grand_total
)

print(
    "\nClean dataset:",
    CLEAN_DIR
)

print(
    "Final split:",
    SPLIT_DIR
)

print(
    "\nRandom seed:",
    SEED
)

print("=" * 75)
```

```python
# ============================================================
# FINAL CROSS-SPLIT LEAKAGE VERIFICATION
# ============================================================

import os
import hashlib
from collections import defaultdict

SPLIT_DIR = "/content/drive/MyDrive/Urban_Dataset_Clean_Split"

SPLITS = [
    "train",
    "validation",
    "test"
]

IMAGE_EXTENSIONS = (
    ".jpg",
    ".jpeg",
    ".png",
    ".bmp",
    ".webp",
    ".avif"
)


def get_image_hash(filepath):

    with open(filepath, "rb") as f:
        return hashlib.md5(f.read()).hexdigest()


# ------------------------------------------------------------
# COLLECT HASHES
# ------------------------------------------------------------

hash_locations = defaultdict(list)

for split in SPLITS:

    split_path = os.path.join(
        SPLIT_DIR,
        split
    )

    for root, dirs, files in os.walk(split_path):

        for filename in files:

            if not filename.lower().endswith(
                IMAGE_EXTENSIONS
            ):
                continue

            filepath = os.path.join(
                root,
                filename
            )

            file_hash = get_image_hash(filepath)

            hash_locations[file_hash].append(
                (split, filepath)
            )


# ------------------------------------------------------------
# FIND CROSS-SPLIT DUPLICATES
# ------------------------------------------------------------

cross_split_duplicates = []

for file_hash, locations in hash_locations.items():

    splits_found = set(
        location[0]
        for location in locations
    )

    if len(splits_found) > 1:

        cross_split_duplicates.append(
            locations
        )


# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("=" * 75)
print("FINAL DATA LEAKAGE VERIFICATION")
print("=" * 75)

print()

print(
    "Total unique image hashes :",
    len(hash_locations)
)

print(
    "Cross-split duplicate groups :",
    len(cross_split_duplicates)
)

print()

if len(cross_split_duplicates) == 0:

    print("=" * 75)
    print("RESULT: NO CROSS-SPLIT DATA LEAKAGE DETECTED")
    print("=" * 75)

    print()
    print("TRAIN, VALIDATION and TEST contain")
    print("no identical image files.")

else:

    print("=" * 75)
    print("WARNING: CROSS-SPLIT DUPLICATES FOUND")
    print("=" * 75)

    for i, group in enumerate(
        cross_split_duplicates,
        start=1
    ):

        print(f"\nDuplicate Group {i}")

        for split, filepath in group:

            print(
                f"[{split.upper()}] {filepath}"
            )
```

```python
# ============================================================
# STEP 11 — COMMON CLASSIFICATION HEAD
# ============================================================

import tensorflow as tf

from tensorflow.keras import layers, Model


def build_transfer_model(
    backbone,
    preprocess_layer,
    model_name
):

    # --------------------------------------------------------
    # Freeze pretrained backbone initially
    # --------------------------------------------------------

    backbone.trainable = False

    # --------------------------------------------------------
    # Input
    # --------------------------------------------------------

    inputs = layers.Input(
        shape=(300, 300, 3),
        name="input_image"
    )

    # --------------------------------------------------------
    # Data augmentation
    # --------------------------------------------------------

    x = data_augmentation(
        inputs,
        training=True
    )

    # --------------------------------------------------------
    # Model-specific preprocessing
    # --------------------------------------------------------

    x = preprocess_layer(x)

    # --------------------------------------------------------
    # Pretrained feature extractor
    # --------------------------------------------------------

    x = backbone(
        x,
        training=False
    )

    # --------------------------------------------------------
    # Global Average Pooling
    # --------------------------------------------------------

    x = layers.GlobalAveragePooling2D(
        name="global_average_pooling"
    )(x)

    # --------------------------------------------------------
    # Classification head
    # --------------------------------------------------------

    x = layers.Dense(
        256,
        activation="relu",
        name="dense_256"
    )(x)

    x = layers.BatchNormalization(
        name="batch_normalization"
    )(x)

    x = layers.Dropout(
        0.30,
        name="dropout"
    )(x)

    # --------------------------------------------------------
    # Four-class output
    # --------------------------------------------------------

    outputs = layers.Dense(
        4,
        activation="softmax",
        name="classification_output"
    )(x)

    # --------------------------------------------------------
    # Final model
    # --------------------------------------------------------

    model = Model(
        inputs=inputs,
        outputs=outputs,
        name=model_name
    )

    return model
```

```python
# ============================================================
# STEP 26 — ADVANCED ROC-AUC ANALYSIS
# Bootstrap 95% Confidence Intervals
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import label_binarize
from sklearn.metrics import (
    roc_curve,
    roc_auc_score
)

# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

N_BOOTSTRAP = 2000
RANDOM_SEED = 42

rng = np.random.default_rng(RANDOM_SEED)

# ------------------------------------------------------------
# TRUE LABELS
# ------------------------------------------------------------

y_true_bin = label_binarize(
    y_true,
    classes=np.arange(len(CLASS_NAMES))
)

# ------------------------------------------------------------
# ACTUAL MODEL PROBABILITIES
# ------------------------------------------------------------

roc_models = {
    "MobileNetV2": y_prob_mobile,
    "EfficientNetB0": y_prob,
    "DenseNet121": y_prob_dense
}

# ------------------------------------------------------------
# BOOTSTRAP ROC-AUC CONFIDENCE INTERVAL
# ------------------------------------------------------------

def bootstrap_auc_ci(
    y_true_binary,
    probabilities,
    n_bootstrap=2000,
    seed=42
):

    rng_local = np.random.default_rng(seed)

    n_samples = len(y_true_binary)

    bootstrap_scores = []

    for _ in range(n_bootstrap):

        indices = rng_local.integers(
            0,
            n_samples,
            n_samples
        )

        y_sample = y_true_binary[indices]
        p_sample = probabilities[indices]

        # Skip samples containing only one class
        if len(np.unique(y_sample)) < 2:
            continue

        score = roc_auc_score(
            y_sample,
            p_sample
        )

        bootstrap_scores.append(score)

    lower = np.percentile(
        bootstrap_scores,
        2.5
    )

    upper = np.percentile(
        bootstrap_scores,
        97.5
    )

    return lower, upper


# ------------------------------------------------------------
# ANALYSIS TABLE
# ------------------------------------------------------------

roc_summary = []

for model_name, probabilities in roc_models.items():

    # Actual multiclass macro ROC-AUC
    macro_auc = roc_auc_score(
        y_true_bin,
        probabilities,
        average="macro",
        multi_class="ovr"
    )

    weighted_auc = roc_auc_score(
        y_true_bin,
        probabilities,
        average="weighted",
        multi_class="ovr"
    )

    # Class-wise bootstrap AUC
    class_auc = []

    for class_idx in range(len(CLASS_NAMES)):

        auc_value = roc_auc_score(
            y_true_bin[:, class_idx],
            probabilities[:, class_idx]
        )

        class_auc.append(auc_value)

    # Bootstrap macro AUC
    bootstrap_macro_scores = []

    for _ in range(N_BOOTSTRAP):

        indices = rng.integers(
            0,
            len(y_true),
            len(y_true)
        )

        y_sample = y_true_bin[indices]
        p_sample = probabilities[indices]

        try:

            score = roc_auc_score(
                y_sample,
                p_sample,
                average="macro",
                multi_class="ovr"
            )

            bootstrap_macro_scores.append(score)

        except ValueError:
            continue

    ci_lower = np.percentile(
        bootstrap_macro_scores,
        2.5
    )

    ci_upper = np.percentile(
        bootstrap_macro_scores,
        97.5
    )

    roc_summary.append({

        "Model": model_name,

        "ROC-AUC Macro": macro_auc,

        "95% CI Lower": ci_lower,

        "95% CI Upper": ci_upper,

        "ROC-AUC Weighted": weighted_auc,

        "Residential AUC": class_auc[0],

        "Commercial AUC": class_auc[1],

        "Construction AUC": class_auc[2],

        "School AUC": class_auc[3]
    })


roc_summary_df = pd.DataFrame(
    roc_summary
)

roc_summary_df = roc_summary_df.sort_values(
    "ROC-AUC Macro",
    ascending=False
).reset_index(drop=True)


# ------------------------------------------------------------
# PRINT RESULTS
# ------------------------------------------------------------

print("=" * 100)
print("ADVANCED ROC-AUC ANALYSIS")
print("=" * 100)

display(
    roc_summary_df.style.format({
        "ROC-AUC Macro": "{:.4f}",
        "95% CI Lower": "{:.4f}",
        "95% CI Upper": "{:.4f}",
        "ROC-AUC Weighted": "{:.4f}",
        "Residential AUC": "{:.4f}",
        "Commercial AUC": "{:.4f}",
        "Construction AUC": "{:.4f}",
        "School AUC": "{:.4f}"
    })
)


# ============================================================
# PUBLICATION FIGURE
# ============================================================

plt.figure(
    figsize=(9, 7)
)

for model_name, probabilities in roc_models.items():

    # --------------------------------------------------------
    # Class-wise ROC curves
    # --------------------------------------------------------

    fpr = {}
    tpr = {}

    for class_idx in range(len(CLASS_NAMES)):

        fpr[class_idx], tpr[class_idx], _ = roc_curve(
            y_true_bin[:, class_idx],
            probabilities[:, class_idx]
        )

    # --------------------------------------------------------
    # Macro-average ROC curve
    # --------------------------------------------------------

    all_fpr = np.unique(
        np.concatenate([
            fpr[i]
            for i in range(len(CLASS_NAMES))
        ])
    )

    mean_tpr = np.zeros_like(
        all_fpr
    )

    for class_idx in range(len(CLASS_NAMES)):

        mean_tpr += np.interp(
            all_fpr,
            fpr[class_idx],
            tpr[class_idx]
        )

    mean_tpr /= len(CLASS_NAMES)

    # IMPORTANT:
    # Use sklearn ROC-AUC value for the label,
    # not auc(all_fpr, mean_tpr)
    macro_auc = roc_auc_score(
        y_true_bin,
        probabilities,
        average="macro",
        multi_class="ovr"
    )

    # --------------------------------------------------------
    # Confidence interval
    # --------------------------------------------------------

    row = roc_summary_df[
        roc_summary_df["Model"] == model_name
    ].iloc[0]

    ci_low = row["95% CI Lower"]
    ci_high = row["95% CI Upper"]

    plt.plot(
        all_fpr,
        mean_tpr,
        linewidth=2.5,
        label=(
            f"{model_name}  "
            f"AUC={macro_auc:.4f} "
            f"(95% CI {ci_low:.4f}–{ci_high:.4f})"
        )
    )


# ------------------------------------------------------------
# RANDOM BASELINE
# ------------------------------------------------------------

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.3,
    label="Random classifier"
)


# ------------------------------------------------------------
# AXES
# ------------------------------------------------------------

plt.xlabel(
    "False Positive Rate",
    fontsize=14
)

plt.ylabel(
    "True Positive Rate",
    fontsize=14
)

plt.title(
    "Macro-Averaged Receiver Operating Characteristic Analysis",
    fontsize=15,
    fontweight="bold",
    pad=15
)

plt.xlim(
    0,
    1
)

plt.ylim(
    0,
    1.02
)

plt.xticks(
    fontsize=11
)

plt.yticks(
    fontsize=11
)

plt.grid(
    alpha=0.20,
    linestyle=":"
)

plt.legend(
    loc="lower right",
    fontsize=9,
    frameon=True
)

plt.tight_layout()


# ------------------------------------------------------------
# SAVE 600 DPI
# ------------------------------------------------------------

output_path = (
    "/content/drive/MyDrive/"
    "Figure_Advanced_ROC_AUC_Analysis.png"
)

plt.savefig(
    output_path,
    dpi=600,
    bbox_inches="tight"
)

plt.show()

print("\n✓ Advanced ROC figure saved:")
print(output_path)
```

```python
# ============================================================
# STEP 28 — PAIRED STATISTICAL MODEL COMPARISON
# McNemar Test + Bootstrap Accuracy Difference
# ============================================================

import numpy as np
import pandas as pd

from math import comb

# ------------------------------------------------------------
# MODEL PREDICTIONS
# ------------------------------------------------------------

model_predictions = {
    "MobileNetV2": y_pred_mobile,
    "EfficientNetB0": y_pred,
    "DenseNet121": y_pred_dense
}

model_names = list(model_predictions.keys())

# ------------------------------------------------------------
# ACTUAL CORRECTNESS
# ------------------------------------------------------------

correctness = {}

for model_name, predictions in model_predictions.items():

    correctness[model_name] = (
        predictions == y_true
    )

# ------------------------------------------------------------
# EXACT MCNEMAR P-VALUE
# ------------------------------------------------------------

def exact_mcnemar_pvalue(
    correct_a,
    correct_b
):

    # A correct, B incorrect
    b = np.sum(
        correct_a & ~correct_b
    )

    # A incorrect, B correct
    c = np.sum(
        ~correct_a & correct_b
    )

    discordant = b + c

    if discordant == 0:
        return 1.0, b, c

    # Exact two-sided binomial test
    smaller = min(b, c)

    probability = (
        sum(
            comb(discordant, k)
            for k in range(smaller + 1)
        )
        / (2 ** discordant)
    )

    p_value = min(
        1.0,
        2 * probability
    )

    return p_value, b, c


# ------------------------------------------------------------
# BOOTSTRAP ACCURACY DIFFERENCE
# ------------------------------------------------------------

def bootstrap_accuracy_difference(
    y_true,
    pred_a,
    pred_b,
    n_bootstrap=5000,
    seed=42
):

    rng = np.random.default_rng(seed)

    n = len(y_true)

    differences = []

    for _ in range(n_bootstrap):

        indices = rng.integers(
            0,
            n,
            n
        )

        acc_a = np.mean(
            pred_a[indices] ==
            y_true[indices]
        )

        acc_b = np.mean(
            pred_b[indices] ==
            y_true[indices]
        )

        differences.append(
            acc_a - acc_b
        )

    differences = np.array(
        differences
    )

    lower = np.percentile(
        differences,
        2.5
    )

    upper = np.percentile(
        differences,
        97.5
    )

    observed = (
        np.mean(pred_a == y_true)
        -
        np.mean(pred_b == y_true)
    )

    return (
        observed,
        lower,
        upper
    )


# ------------------------------------------------------------
# PAIRWISE COMPARISON
# ------------------------------------------------------------

comparison_results = []

for i in range(
    len(model_names)
):

    for j in range(
        i + 1,
        len(model_names)
    ):

        model_a = model_names[i]
        model_b = model_names[j]

        pred_a = np.asarray(
            model_predictions[model_a]
        )

        pred_b = np.asarray(
            model_predictions[model_b]
        )

        # McNemar
        p_value, b, c = exact_mcnemar_pvalue(
            correctness[model_a],
            correctness[model_b]
        )

        # Bootstrap accuracy difference
        observed_diff, ci_lower, ci_upper = (
            bootstrap_accuracy_difference(
                np.asarray(y_true),
                pred_a,
                pred_b
            )
        )

        accuracy_a = np.mean(
            pred_a == y_true
        )

        accuracy_b = np.mean(
            pred_b == y_true
        )

        comparison_results.append({

            "Model A": model_a,

            "Model B": model_b,

            "Accuracy A": accuracy_a,

            "Accuracy B": accuracy_b,

            "Difference A-B": observed_diff,

            "95% CI Lower": ci_lower,

            "95% CI Upper": ci_upper,

            "McNemar b": b,

            "McNemar c": c,

            "McNemar p-value": p_value
        })


comparison_df = pd.DataFrame(
    comparison_results
)


# ------------------------------------------------------------
# DISPLAY
# ------------------------------------------------------------

print("=" * 110)
print("PAIRED STATISTICAL MODEL COMPARISON")
print("=" * 110)

display(
    comparison_df.style.format({

        "Accuracy A": "{:.4f}",

        "Accuracy B": "{:.4f}",

        "Difference A-B": "{:.4f}",

        "95% CI Lower": "{:.4f}",

        "95% CI Upper": "{:.4f}",

        "McNemar p-value": "{:.6f}"
    })
)


# ------------------------------------------------------------
# INTERPRETATION
# ------------------------------------------------------------

print("\n")
print("=" * 110)
print("STATISTICAL INTERPRETATION")
print("=" * 110)

for _, row in comparison_df.iterrows():

    print(
        f"\n{row['Model A']} vs {row['Model B']}"
    )

    print(
        f"Accuracy difference: "
        f"{row['Difference A-B']:.4f}"
    )

    print(
        f"95% bootstrap CI: "
        f"[{row['95% CI Lower']:.4f}, "
        f"{row['95% CI Upper']:.4f}]"
    )

    print(
        f"McNemar p-value: "
        f"{row['McNemar p-value']:.6f}"
    )

    if row["McNemar p-value"] < 0.05:

        print(
            "Result: Statistically significant "
            "difference (p < 0.05)"
        )

    else:

        print(
            "Result: No statistically significant "
            "difference (p >= 0.05)"
        )
```

```python
# ============================================================
# STEP 30 — CROSS-MODEL ERROR TRANSITION ANALYSIS
# ============================================================

import pandas as pd
import numpy as np

model_predictions = {
    "MobileNetV2": y_pred_mobile,
    "EfficientNetB0": y_pred,
    "DenseNet121": y_pred_dense
}

error_rows = []

for model_name, predictions in model_predictions.items():

    cm = confusion_matrix(
        y_true,
        predictions
    )

    for actual_idx, actual_class in enumerate(CLASS_NAMES):

        for predicted_idx, predicted_class in enumerate(CLASS_NAMES):

            # Ignore correct predictions
            if actual_idx == predicted_idx:
                continue

            count = cm[
                actual_idx,
                predicted_idx
            ]

            if count > 0:

                error_rows.append({
                    "Model": model_name,
                    "Actual Class": actual_class,
                    "Predicted Class": predicted_class,
                    "Errors": int(count),
                    "Error Rate Within Class": (
                        count /
                        cm[actual_idx].sum()
                    )
                })


error_df = pd.DataFrame(
    error_rows
)

error_df = error_df.sort_values(
    ["Model", "Errors"],
    ascending=[True, False]
).reset_index(drop=True)


print("=" * 100)
print("CROSS-MODEL ERROR TRANSITION ANALYSIS")
print("=" * 100)

display(
    error_df.style.format({
        "Error Rate Within Class": "{:.2%}"
    })
)


# ------------------------------------------------------------
# TOP ERROR PATTERN FOR EACH MODEL
# ------------------------------------------------------------

print("\n")
print("=" * 100)
print("DOMINANT ERROR PATTERN BY MODEL")
print("=" * 100)

for model_name in model_predictions.keys():

    model_errors = error_df[
        error_df["Model"] == model_name
    ]

    top_error = model_errors.iloc[0]

    print(
        f"\n{model_name}: "
        f"{top_error['Actual Class']} → "
        f"{top_error['Predicted Class']} "
        f"({top_error['Errors']} images, "
        f"{top_error['Error Rate Within Class']:.2%} "
        f"of that actual class)"
    )


# ------------------------------------------------------------
# AGGREGATED ERROR TRANSITIONS
# ------------------------------------------------------------

transition_summary = (
    error_df
    .groupby(
        ["Actual Class", "Predicted Class"],
        as_index=False
    )["Errors"]
    .sum()
    .sort_values(
        "Errors",
        ascending=False
    )
)

print("\n")
print("=" * 100)
print("AGGREGATED ERROR TRANSITIONS ACROSS ALL MODELS")
print("=" * 100)

display(
    transition_summary
)
```

```python
# ============================================================
# STEP 31 — CONFIDENCE-BASED ERROR ANALYSIS
# ============================================================

import numpy as np
import pandas as pd

model_probabilities = {
    "MobileNetV2": y_prob_mobile,
    "EfficientNetB0": y_prob,
    "DenseNet121": y_prob_dense
}

confidence_error_rows = []

for model_name, probabilities in model_probabilities.items():

    predictions = np.argmax(
        probabilities,
        axis=1
    )

    for i in range(len(y_true)):

        # Only analyze incorrect predictions
        if predictions[i] != y_true[i]:

            sorted_probs = np.sort(
                probabilities[i]
            )[::-1]

            top1_confidence = sorted_probs[0]
            top2_confidence = sorted_probs[1]

            confidence_margin = (
                top1_confidence -
                top2_confidence
            )

            actual_probability = probabilities[
                i,
                y_true[i]
            ]

            confidence_error_rows.append({

                "Model": model_name,

                "Sample Index": i,

                "Actual Class":
                    CLASS_NAMES[y_true[i]],

                "Predicted Class":
                    CLASS_NAMES[predictions[i]],

                "Predicted Confidence":
                    top1_confidence,

                "Actual-Class Probability":
                    actual_probability,

                "Confidence Margin":
                    confidence_margin
            })


confidence_error_df = pd.DataFrame(
    confidence_error_rows
)


# ------------------------------------------------------------
# SORT BY HIGH-CONFIDENCE ERRORS
# ------------------------------------------------------------

high_confidence_errors = (
    confidence_error_df
    .sort_values(
        "Predicted Confidence",
        ascending=False
    )
    .reset_index(drop=True)
)


print("=" * 110)
print("HIGH-CONFIDENCE MISCLASSIFICATIONS")
print("=" * 110)

display(
    high_confidence_errors.head(20).style.format({

        "Predicted Confidence":
            "{:.4f}",

        "Actual-Class Probability":
            "{:.4f}",

        "Confidence Margin":
            "{:.4f}"
    })
)


# ------------------------------------------------------------
# LOW-CONFIDENCE ERRORS
# ------------------------------------------------------------

low_confidence_errors = (
    confidence_error_df
    .sort_values(
        "Predicted Confidence",
        ascending=True
    )
    .reset_index(drop=True)
)


print("\n")
print("=" * 110)
print("LOW-CONFIDENCE MISCLASSIFICATIONS")
print("=" * 110)

display(
    low_confidence_errors.head(20).style.format({

        "Predicted Confidence":
            "{:.4f}",

        "Actual-Class Probability":
            "{:.4f}",

        "Confidence Margin":
            "{:.4f}"
    })
)


# ------------------------------------------------------------
# AVERAGE ERROR CONFIDENCE BY MODEL
# ------------------------------------------------------------

confidence_summary = (
    confidence_error_df
    .groupby("Model")
    .agg(
        Mean_Error_Confidence=(
            "Predicted Confidence",
            "mean"
        ),

        Mean_Actual_Class_Probability=(
            "Actual-Class Probability",
            "mean"
        ),

        Mean_Confidence_Margin=(
            "Confidence Margin",
            "mean"
        )
    )
    .reset_index()
)


print("\n")
print("=" * 110)
print("ERROR CONFIDENCE SUMMARY")
print("=" * 110)

display(
    confidence_summary.style.format({

        "Mean_Error_Confidence":
            "{:.4f}",

        "Mean_Actual_Class_Probability":
            "{:.4f}",

        "Mean_Confidence_Margin":
            "{:.4f}"
    })
)
```

```python
# ============================================================
# STEP 32 — MAP TEST INDICES TO ACTUAL IMAGE FILES
# ============================================================

# Reconstruct test file paths in the SAME ORDER as test_dataset
test_file_paths = []

for class_name in CLASS_NAMES:

    class_dir = os.path.join(
        TEST_DIR,
        class_name
    )

    for filename in sorted(
        os.listdir(class_dir)
    ):

        filepath = os.path.join(
            class_dir,
            filename
        )

        if os.path.isfile(filepath):

            test_file_paths.append(
                filepath
            )


# ------------------------------------------------------------
# VERIFY ORDER
# ------------------------------------------------------------

print("Number of test paths:", len(test_file_paths))
print("Number of true labels:", len(y_true))

assert len(test_file_paths) == len(y_true), (
    "Test path count does not match prediction count."
)


# ------------------------------------------------------------
# HIGH-CONFIDENCE MOBILE NET ERRORS
# ------------------------------------------------------------

mobile_errors = confidence_error_df[
    confidence_error_df["Model"] == "MobileNetV2"
].copy()

mobile_errors = mobile_errors.sort_values(
    "Predicted Confidence",
    ascending=False
).reset_index(drop=True)


mobile_errors["Image Path"] = (
    mobile_errors["Sample Index"]
    .apply(
        lambda x: test_file_paths[int(x)]
    )
)


print("=" * 110)
print("MOBILENETV2 — HIGH-CONFIDENCE ERROR IMAGES")
print("=" * 110)

display(
    mobile_errors[
        [
            "Sample Index",
            "Actual Class",
            "Predicted Class",
            "Predicted Confidence",
            "Actual-Class Probability",
            "Confidence Margin",
            "Image Path"
        ]
    ].head(15).style.format({

        "Predicted Confidence": "{:.4f}",

        "Actual-Class Probability": "{:.4f}",

        "Confidence Margin": "{:.4f}"
    })
)
```
